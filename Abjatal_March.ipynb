```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("/content/ABJATAL DAILY REPORT 2025 _ 2026.xlsx")

# Set the first row as column headers and then drop it from the data
df.columns = df.iloc[0]
df = df[1:].reset_index(drop=True)

df.drop_duplicates(inplace=True)
# Removed extraneous df.head(), df.info(), df.describe(), df, df.dropna() calls
```

```python
df_clean = df.dropna(axis=1)  # Or axis='columns'
df
```

```python
df = df[~(df == 0).all(axis=1)]
df
```

```python
df = df.dropna(how='all')
df
```

```python
# Convert 'NUMBER' column to numeric, coercing errors to NaN
df['NUMBER'] = pd.to_numeric(df['NUMBER'], errors='coerce')

# Drop rows where 'NUMBER' is 0
df = df[df['NUMBER'] != 0]

# Display the first few rows of the updated DataFrame
print(df.head())
```

```python
# Select all columns named 'OM CASH IN'
cash_in_df_to_convert = df.filter(like='OM CASH IN')

# Select all columns named 'OM CASH OUT'
cash_out_df_to_convert = df.filter(like='OM CASH OUT')

# Convert all columns in these filtered DataFrames to numeric
cash_in_df_numeric = cash_in_df_to_convert.apply(pd.to_numeric, errors='coerce')
cash_out_df_numeric = cash_out_df_to_convert.apply(pd.to_numeric, errors='coerce')

# Sum the numeric cash in/out columns
df['Total_Cash_In'] = cash_in_df_numeric.sum(axis=1)
df['Total_Cash_Out'] = cash_out_df_numeric.sum(axis=1)

# Ensure 'NUMBER' is numeric before grouping
df['NUMBER'] = pd.to_numeric(df['NUMBER'], errors='coerce')

# Group by 'NUMBER' (Agent Number) and sum the calculated totals
agent_summary = df.groupby('NUMBER')[['Total_Cash_In', 'Total_Cash_Out']].sum().reset_index()

# Drop any NaN 'NUMBER' entries that might have resulted from coercion
agent_summary.dropna(subset=['NUMBER'], inplace=True)

# Get top 20 agents by Total Cash In
top_20_cash_in = agent_summary.nlargest(20, 'Total_Cash_In')

# Get top 20 agents by Total Cash Out
top_20_cash_out = agent_summary.nlargest(20, 'Total_Cash_Out')

print("Top 20 Agents by Total Cash In:")
display(top_20_cash_in)

print("\nTop 20 Agents by Total Cash Out:")
display(top_20_cash_out)
```

```python
# Get top 10 agents by Total Cash In
top_10_cash_in = agent_summary.nlargest(10, 'Total_Cash_In')

print("Top 10 Agents by Total Cash In:")
display(top_10_cash_in)
```

```python
# Get top 10 agents by Total Cash Out
top_10_cash_out = agent_summary.nlargest(10, 'Total_Cash_Out')

print("\nTop 10 Agents by Total Cash Out:")
display(top_10_cash_out)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# --- Plot for Top 10 Agents by Total Cash In ---
plt.figure(figsize=(12, 7))
sns.barplot(x='NUMBER', y='Total_Cash_In', data=top_10_cash_in, color='#4CAF50', label='Total Cash In')
sns.barplot(x='NUMBER', y='Total_Cash_Out', data=top_10_cash_in, color='#FF5722', alpha=0.8, label='Total Cash Out')
plt.title('Top 10 Agents by Total Cash In and Corresponding Cash Out')
plt.xlabel('Agent Number')
plt.ylabel('Amount')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# - Plot for Top 10 Agents by Total Cash Out
plt.figure(figsize=(12, 7))
sns.barplot(x='NUMBER', y='Total_Cash_Out', data=top_10_cash_out, color='#FF5722', label='Total Cash Out')
sns.barplot(x='NUMBER', y='Total_Cash_In', data=top_10_cash_out, color='#4CAF50', alpha=0.8, label='Total Cash In')
plt.title('Top 10 Agents by Total Cash Out and Corresponding Cash In')
plt.xlabel('Agent Number')
plt.ylabel('Amount')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

### Agent Performance: Net Cash Flow

```python
# Calculate Net Cash Flow for each agent
agent_summary['Net_Cash_Flow'] = agent_summary['Total_Cash_In'] - agent_summary['Total_Cash_Out']

# Sort agents by Net Cash Flow (highest to lowest)
agent_net_cash_flow_sorted = agent_summary.sort_values(by='Net_Cash_Flow', ascending=False)

print("Top 10 Agents by Net Cash Flow (Positive):")
display(agent_net_cash_flow_sorted.head(10))

print("\nBottom 10 Agents by Net Cash Flow (Negative):")
display(agent_net_cash_flow_sorted.tail(10))
```

### Visualization: Top and Bottom Net Cash Flow Agents

```python
# Visualize Top 10 Agents by Net Cash Flow
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Net_Cash_Flow', data=agent_net_cash_flow_sorted.head(10), palette='viridis', hue='NUMBER', legend=False)
plt.title('Top 10 Agents by Net Cash Flow')
plt.xlabel('Agent Number')
plt.ylabel('Net Cash Flow')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Visualize Bottom 10 Agents by Net Cash Flow
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Net_Cash_Flow', data=agent_net_cash_flow_sorted.tail(10), palette='magma', hue='NUMBER', legend=False)
plt.title('Bottom 10 Agents by Net Cash Flow')
plt.xlabel('Agent Number')
plt.ylabel('Net Cash Flow')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Agent Performance: Maximum and Lowest Bonuses

```python
# Extract the 'BONUS' column from the original df and convert to numeric
# We need to find the column that corresponds to 'BONUS' in the header row.
# Assuming 'BONUS' is a unique string in the header row, we can locate it.
bonus_cols = df.filter(like='BONUS', axis=1)

# Assuming there's only one 'BONUS' column or we sum them if there are multiple.
# Let's verify this first. If not unique, we'll need to adapt.
if bonus_cols.shape[1] > 1:
    print("Warning: Multiple 'BONUS' columns found. Summing them up.")
    df['Total_Bonus'] = bonus_cols.apply(pd.to_numeric, errors='coerce').sum(axis=1)
else:
    df['Total_Bonus'] = pd.to_numeric(bonus_cols.iloc[:, 0], errors='coerce')

# Group by 'NUMBER' and sum the total bonus
agent_bonus_summary = df.groupby('NUMBER')['Total_Bonus'].sum().reset_index()

# Merge Total_Bonus into agent_summary
agent_summary = pd.merge(agent_summary, agent_bonus_summary, on='NUMBER', how='left')

# Handle potential NaN values in 'Total_Bonus' after merge (e.g., agents with no bonus data)
agent_summary['Total_Bonus'] = agent_summary['Total_Bonus'].fillna(0)

# Get agents with the maximum bonuses
max_bonus_agents = agent_summary.nlargest(10, 'Total_Bonus')

# Get agents with the lowest (could be zero or negative if any) bonuses
# Considering lowest bonus means smallest positive or largest negative. If only 0, then just sorting asc will work.
min_bonus_agents = agent_summary.nsmallest(10, 'Total_Bonus')

print("Top 10 Agents by Total Bonus:")
display(max_bonus_agents)

print("\nBottom 10 Agents by Total Bonus:")
display(min_bonus_agents)
```

### Visualization: Top and Bottom Bonus Agents

```python
# Visualize Top 10 Agents by Total Bonus
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Total_Bonus', data=max_bonus_agents, palette='Greens_r', hue='NUMBER', legend=False)
plt.title('Top 10 Agents by Total Bonus')
plt.xlabel('Agent Number')
plt.ylabel('Total Bonus')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Visualize Bottom 10 Agents by Total Bonus
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Total_Bonus', data=min_bonus_agents, palette='Reds_r', hue='NUMBER', legend=False)
plt.title('Bottom 10 Agents by Total Bonus')
plt.xlabel('Agent Number')
plt.ylabel('Total Bonus')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Agent Performance: Total Commission

```python
# Extract the 'AGENT COMM.' column from the original df and convert to numeric
# Assuming 'AGENT COMM.' is a unique string in the header row.
comm_cols = df.filter(like='AGENT COMM.', axis=1)

# Assuming there's only one 'AGENT COMM.' column or we sum them if there are multiple.
if comm_cols.shape[1] > 1:
    print("Warning: Multiple 'AGENT COMM.' columns found. Summing them up.")
    df['Total_Commission'] = comm_cols.apply(pd.to_numeric, errors='coerce').sum(axis=1)
else:
    df['Total_Commission'] = pd.to_numeric(comm_cols.iloc[:, 0], errors='coerce')

# Group by 'NUMBER' and sum the total commission
agent_comm_summary = df.groupby('NUMBER')['Total_Commission'].sum().reset_index()

# Merge Total_Commission into agent_summary
agent_summary = pd.merge(agent_summary, agent_comm_summary, on='NUMBER', how='left')

# Handle potential NaN values in 'Total_Commission' after merge (e.g., agents with no commission data)
agent_summary['Total_Commission'] = agent_summary['Total_Commission'].fillna(0)

# Get agents with the maximum commissions
max_comm_agents = agent_summary.nlargest(10, 'Total_Commission')

# Get agents with the lowest commissions
min_comm_agents = agent_summary.nsmallest(10, 'Total_Commission')

print("Top 10 Agents by Total Commission:")
display(max_comm_agents)

print("\nBottom 10 Agents by Total Commission:")
display(min_comm_agents)
```

### Visualization: Top and Bottom Commission Agents

```python
# Visualize Top 10 Agents by Total Commission
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Total_Commission', data=max_comm_agents, palette='YlGnBu_r', hue='NUMBER', legend=False)
plt.title('Top 10 Agents by Total Commission')
plt.xlabel('Agent Number')
plt.ylabel('Total Commission')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Visualize Bottom 10 Agents by Total Commission
plt.figure(figsize=(14, 7))
sns.barplot(x='NUMBER', y='Total_Commission', data=min_comm_agents, palette='PuRd_r', hue='NUMBER', legend=False)
plt.title('Bottom 10 Agents by Total Commission')
plt.xlabel('Agent Number')
plt.ylabel('Total Commission')
plt.ticklabel_format(style='plain', axis='y')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
